```python
import matplotlib.pyplot as plt
import numpy as np
from pyBraytonHeatPump import BraytonHeatPump  # 替换为实际模块名

# 初始化固定参数
base_params = {
    "flue_gas_composition": {
        "co2": 0.1338,
        "o2": 0.0384,
        "n2": 0.6975
    },
    "isentropic_eff_mc": 0.88,
    "t_isentropic_eff_mc": 0.92,
    "mechanical_eff": 0.98,
    "industrial_waste_heat_t": 300,
    "heat_transfer_loss_eff": 0.96,
    "t_amb": 20,
    "p_amb": 101325,
    "p_bray_L": 7.5e6
}

base_inputs = {
    "p_bray_H": 19447839.26865841,
    "p_bray_M": 12827110.4341202,
    "p_Dehy": 1e5,
    "Compressor power limit": 200e6,
    "Turbine power limit": 35e6,
    "Dehy_overheating_temperature": 20,
    "min_temperature_exchange": 15
}

# 生成功率范围 (单位：W)
power_range = np.linspace(10e6, 200e6, 20)  # 在10-200MW之间生成20个点

# 存储结果的字典
results_data = {
    "Model1": [],
    "Model2": []
}

# 进行参数扫描
for econ_model in [1, 2]:
    print(f"\n正在计算经济模型 {econ_model}...")
    
    for power in power_range:
        # 复制基础参数避免污染
        current_params = base_params.copy()
        current_inputs = base_inputs.copy()
        
        # 更新变量参数
        current_params["Store_electrical_power"] = power
        current_inputs["Economic Model Selection"] = econ_model
        
        # 创建并求解模型
        try:
            brayton = BraytonHeatPump(current_params)
            results = brayton.solve(current_inputs)
            
            # 提取换热器成本（处理可能的单位转换）
            #cost = results["cost"]["cost_exchanger"]
            cost = results["cost"]["cost_hr"]
            results_data[f"Model{econ_model}"].append(cost/1e3)  # 转换为千美元
            
        except Exception as e:
            print(f"在{power/1e6}MW时出错: {str(e)}")
            results_data[f"Model{econ_model}"].append(np.nan)

# 数据可视化
plt.figure(figsize=(12, 7), dpi=100)
x_axis = power_range / 1e6  # 转换为MW单位

# 绘制模型1曲线
plt.plot(x_axis, results_data["Model1"], 
         'o-', 
         linewidth=2,
         markersize=6,
         color='#2C5F2D',
         label='Tesio 模型')

# 绘制模型2曲线
plt.plot(x_axis, results_data["Model2"], 
         's--', 
         linewidth=2,
         markersize=6,
         color='#E14658',
         label='Nathan T 模型')

# 图表装饰
plt.title("换热器成本 vs 存储功率", fontsize=14, pad=20)
plt.xlabel("存储功率 (MW)", fontsize=12)
plt.ylabel("换热器成本 (千美元)", fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend(frameon=True, fontsize=10)

# 优化显示范围
max_cost = max(max(results_data["Model1"]), max(results_data["Model2"]))
plt.ylim(0, max(max_cost)*1.1)

# 显示图表
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from pyBrayton import Brayton  # 替换为实际模块名

# 初始化固定参数（与.py文件保持一致）
base_params = {
    "flue_gas_composition": {
        "co2": 0.1338,
        "o2": 0.0384,
        "n2": 0.6975
    },
    "isentropic_eff_mc": 0.88,
    "t_isentropic_eff_mc": 0.92,
    "mechanical_eff": 0.98,
    "industrial_waste_heat_t": 300,
    "heat_transfer_loss_eff": 0.96,
    "t_amb": 20,
    "p_amb": 101325,
    "hydrator_eff": 0.95,
    "p_bray_L_B": 7.5e6
}

# 固定输入参数（与.py文件保持一致）
base_inputs = {
    "p_bray_H_B": 28e6,
    "p_bray_MH_B": 16217752.142109105,
    "p_bray_ML_B": 12217752.142109105,
    "p_Hydr": 1e5,
    "Hydr_overheating_temperature": 40,
    "min_temperature_exchange": 15,
    "Compressor power limit": 200e6,
    "Turbine power limit": 35e6
}

# 生成测试范围 (单位：W)
heat_range = np.linspace(10e6, 200e6, 20)  # 在10-200MW之间生成20个点

# 存储结果的字典
results_data = {
    "Model1": [],
    "Model2": []
}

# 进行参数扫描
for econ_model in [1, 2]:
    print(f"\n正在计算经济模型 {econ_model}...")
    
    for heat_value in heat_range:
        # 复制基础配置避免污染
        current_params = base_params.copy()
        current_inputs = base_inputs.copy()
        
        # 更新变量参数
        current_inputs["Economic Model Selection"] = econ_model
        
        try:
            # 初始化模型
            brayton = Brayton(current_params)
            
            # 调用求解器（注意参数顺序！）
            results = brayton.solve(heat_value, current_inputs)
            
            # 提取换热器成本（处理单位转换）
            cost = results["cost"]["cost_exchanger"]
            results_data[f"Model{econ_model}"].append(cost/1e6)  # 转换为百万美元
            
        except Exception as e:
            print(f"在{heat_value/1e6:.1f}MW时出错: {str(e)}")
            results_data[f"Model{econ_model}"].append(np.nan)

# 数据可视化
plt.figure(figsize=(12, 7), dpi=100)
x_axis = heat_range / 1e6  # 转换为MW单位

# 样式配置
style_config = {
    "Model1": {"color": "#1f77b4", "marker": "o", "label": "Tesio 模型"},
    "Model2": {"color": "#ff7f0e", "marker": "s", "label": "Nathan T 模型"}
}

# 绘制双曲线
for model in ["Model1", "Model2"]:
    plt.plot(x_axis, results_data[model],
            marker=style_config[model]["marker"],
            linestyle='--',
            linewidth=1.5,
            color=style_config[model]["color"],
            label=style_config[model]["label"])

# 图表装饰
plt.title("换热器成本 vs 水合器热负荷", fontsize=14, pad=20)
plt.xlabel("水合器热负荷 (MW)", fontsize=12)
plt.ylabel("换热器成本 (百万美元)", fontsize=12)
plt.grid(True, alpha=0.3, linestyle=':')
plt.legend(frameon=True, fontsize=10, loc='upper left')

# 自动调整坐标轴范围
valid_data = [x for x in results_data["Model1"] + results_data["Model2"] if not np.isnan(x)]
if valid_data:
    plt.ylim(0.9*min(valid_data), 1.1*max(valid_data))

# 显示图表
plt.tight_layout()
plt.show()
```
